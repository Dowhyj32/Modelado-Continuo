```python
from scipy.stats import norm

# convertir notas a numéricas
notas = pd.to_numeric(d['Nota'], errors='coerce').dropna()

# media y desvío estándar
mu = notas.mean()
sigma = notas.std()

# histograma normalizado
plt.hist(notas, bins=10, density=True)

# valores para dibujar la curva
x = np.linspace(notas.min(), notas.max(), 100)

# curva normal
y = norm.pdf(x, mu, sigma)

# graficar curva
plt.plot(x, y)

plt.xlabel('Nota')
plt.ylabel('Densidad')
plt.title('Histograma con curva normal')
plt.show()
```
