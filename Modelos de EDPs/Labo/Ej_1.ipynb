```python
import numpy as np

def D_forward(n,h):
    """
    Matriz de diferencias forward.

    Devuelve una matriz de tamaño (n-1)x n.
    """
    return (1/h)*(np.diag(-np.ones(n)) + np.diag(np.ones(n-1),1))[:-1,:]

def D_backward(n,h):
    return (1/h)*(np.diag(np.ones(n)) + np.diag(-np.ones(n-1),-1))[1:,:]

def D_centrada(n,h):
    A = (np.diag(np.ones(n-1),1) - np.diag(np.ones(n-1),-1))/(2*h)

    return A[1:-1,:]
```
