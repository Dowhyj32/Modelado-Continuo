```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Función periódica de período 2π
# ---------------------------------------------------
def f(x, d):
    """
    Devuelve la función:
    
        f(x) = 1/d   si |x| < d
               0     en otro caso
    
    extendida periódicamente con período 2π.
    """

    # Lleva x al intervalo (-π, π]
    x_mod = ((x + np.pi) % (2 * np.pi)) - np.pi

    # Construcción de la función
    return np.where(np.abs(x_mod) < d, 1/d, 0)


# ---------------------------------------------------
# Valores de d a graficar
# ---------------------------------------------------
d_values = [-1, -0.5, 1.5, 0.2, 2]

# Dominio para visualizar
x = np.linspace(-3*np.pi, 3*np.pi, 5000)

# ---------------------------------------------------
# Gráficos
# ---------------------------------------------------
plt.figure(figsize=(12, 8))

for d in d_values:
    y = f(x, d)
    plt.plot(x, y, label=f'd = {d}')

# Decoración
plt.title('Función periódica para distintos valores de d')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()

plt.show()
```
