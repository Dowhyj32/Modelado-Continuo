```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    x = np.mod(x, np.pi)   # periodicidad π

    return np.where(
        (0 < x) & (x < np.pi/3),
        np.pi/3,
        np.where(
            (np.pi/3 < x) & (x < 2*np.pi/3),
            0,
            -np.pi/3
        )
    )
```

```python
x = np.linspace(0, np.pi, 1000)
y = f(x)

plt.figure(figsize=(8,4))
plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)

plt.xticks(
    [0, np.pi/3, 2*np.pi/3, np.pi],
    ['0', r'$\pi/3$', r'$2\pi/3$', r'$\pi$']
)

plt.grid(True)
plt.show()
```
