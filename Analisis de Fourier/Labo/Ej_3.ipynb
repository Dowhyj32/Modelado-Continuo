```python
import numpy as np
import matplotlib.pyplot as plt
from funciones import DFT
```

```python
f = [1, 4, 7]
A = [3, 1, 0.5]

freq_s = np.array([10, 100, 1000, 10000])        # frecuencia de sampleo

for fs in freq_s:
    dt = 1/fs       # tiempo entre muestras
    t = np.arange(0, 1, dt)

    onda = (A[0] * np.sin(2*np.pi*f[0]*t)
        + A[1] * np.sin(2*np.pi*f[1]*t) 
        + A[2] * np.sin(2*np.pi*f[2]*t))


    plt.plot(t, onda, label=f'fs: {fs}')
    
plt.title(f'Onda 3 senos')
plt.grid(True)
plt.legend()
plt.show()
```

```python
f = [1, 4, 7]
A = [3, 1, 0.5]

freq_s = np.array([10, 100, 1000, 10000])        # frecuencia de sampleo

for fs in freq_s:
    dt = 1/fs       # tiempo entre muestras
    t = np.arange(0, 1, dt)

    onda = (A[0] * np.sin(2*np.pi*f[0]*t)
        + A[1] * np.sin(2*np.pi*f[1]*t) 
        + A[2] * np.sin(2*np.pi*f[2]*t))

    transformada, tiempo_calculo_transformada = DFT(onda)
    n = np.arange(fs)
    
    
    plt.figure(figsize=(12,4))

    plt.subplot(1, 2, 1)
    plt.plot(t, onda)
    plt.title(f'Onda 3 senos fs: {fs}')
    plt.grid(True)

    freq = np.arange(len(transformada)) * fs / len(transformada)
    magnitud = np.abs(transformada)
    print(f't: {tiempo_calculo_transformada}s')
    
    plt.subplot(1, 2, 2)
    plt.stem(freq, magnitud)
    
    for i in f:
        plt.text(freq[i], magnitud[i]+2, f'{freq[i]:.0f} Hz, {magnitud[i]:.1f}')
    
    plt.title(f'Transformada fs: {fs}s')
    plt.xlabel('Frecuencia [Hz]')
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```
