```python
import numpy as np
import matplotlib.pyplot as plt
from funciones import DFT
```

```python
A = 10

f = 3
w = 2 * np.pi * f

t = np.arange(0, 1, 1/1000)



N = 30
for d in [0.01, 0.1, 0.2, 0.5, 0.7, 0.9, 1]:
    pulsada = A
    for n in range(1, N+1):
        a_n = 2 * A * np.sin(n*np.pi*d)/(n*np.pi)
        pulsada += a_n *  np.cos(n*w*t)


    plt.plot(t, pulsada, label=f'd:{d}')
    plt.title(f'Señal reconstruida con {N} armónicos')

plt.legend()    
plt.show()
```

```python

# Parámetros
A = 10
d = 0.5

f0 = 3
w0 = 2 * np.pi * f0

N = 80                     # cantidad de armónicos
t = np.arange(0, 1, 1/1000)

# ----------------------------
# Pulsada
# ----------------------------

pulsada = A

coeficientes = []

for n in range(1, N + 1):

    # coeficiente de Fourier
    a_n = (2 * A * np.sin(n * np.pi * d)) / (n * np.pi)

    coeficientes.append(a_n)

    # suma de armónicos
    pulsada += a_n * np.cos(n * w0 * t)

# ----------------------------
# Gráficos
# ----------------------------

armonicos = np.arange(1, N + 1)

plt.figure(figsize=(12,5))
plt.title(f'Pulsada')

# Señal reconstruida
plt.subplot(1,2,1)
plt.plot(t, pulsada)
plt.title(f'Señal reconstruida - d={d}')
plt.xlabel('Tiempo [s]')
plt.grid(True)

# Coeficientes de Fourier
plt.subplot(1,2,2)
plt.stem(armonicos, np.abs(coeficientes))
plt.title('Coeficientes de Fourier')
plt.xlabel('Armónico n')
plt.ylabel('|a_n|')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Parámetros
A = 10

f0 = 3
w0 = 2 * np.pi * f0

N = 10                    # cantidad de armónicos
t = np.arange(0, 1, 1/1000)

# ----------------------------
# Diente de sierra
# ----------------------------

diente_sierra = 0

coeficientes = []

for n in range(1, N + 1):

    # coeficiente de Fourier
    b_n = A / (n * np.pi)

    coeficientes.append(a_n)

    # suma de armónicos
    diente_sierra += b_n * np.sin(n * w0 * t)

# ----------------------------
# Gráficos
# ----------------------------

armonicos = np.arange(1, N + 1)

plt.figure(figsize=(12,5))
plt.title(f'Diente de sierra')

# Señal reconstruida
plt.subplot(1,2,1)
plt.plot(t, diente_sierra)
plt.title(f'Señal reconstruida')
plt.xlabel('Tiempo [s]')
plt.grid(True)

# Coeficientes de Fourier
plt.subplot(1,2,2)
plt.stem(armonicos, np.abs(coeficientes))
plt.title('Coeficientes de Fourier')
plt.xlabel('Armónico n')
plt.ylabel('|a_n|')
plt.grid(True)

plt.tight_layout()
plt.show()
```
