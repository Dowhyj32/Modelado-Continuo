```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def DFT(x):
    N = len(x)
    
    # Vector frecuencias discretas
    X = np.zeros(N, dtype=complex)

    for n in range(N):          # índice de frecuencia
        suma = 0

        for k in range(N):      # índice temporal / muestra
            suma += x[k] * np.exp(-1j * 2 * np.pi * n * k / N)

        X[n] = suma

    return X
```

```python
f = [1, 4, 7]
A = [3, 1, 0.5]

freq_s = np.array([14, 28, 50, 100])        # frecuencia de sampleo

for fs in freq_s:
    dt = 1/fs       # tiempo entre muestras
    t = np.arange(0, 1, dt)

    onda = (A[0] * np.sin(2*np.pi*f[0]*t)
        + A[1] * np.sin(2*np.pi*f[1]*t) 
        + A[2] * np.sin(2*np.pi*f[2]*t))


    plt.plot(t, onda, label=f'fs: {fs}')
    
plt.title(f'Onda 3 senos')
plt.grid(True)
plt.legend()
plt.show()
```

```python
f = [1, 4, 7]
A = [3, 1, 0.5]

fs = 100        # frecuencia de sampleo -> cuantas muestras por segundo tomamos
dt = 1/fs       # tiempo entre muestras consecutivas

t = np.arange(0, 1, dt)     # muestras

onda = (A[0] * np.sin(2*np.pi*f[0]*t)
    + A[1] * np.sin(2*np.pi*f[1]*t) 
    + A[2] * np.sin(2*np.pi*f[2]*t))

transformada = DFT(onda)
n = np.arange(fs)

plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.plot(t, onda)
plt.title(f'Onda 3 senos')
plt.grid(True)

plt.subplot(1, 3, 2)
plt.stem(t, onda)
plt.title(f'Onda 3 senos')
plt.grid(True)

plt.subplot(1, 3, 3)
plt.stem(n, np.abs(transformada))
plt.title(f'Transformada')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
f = [1, 4, 7]
A = [3, 1, 0.5]

fs = 100        # frecuencia de sampleo -> cuantas muestras por segundo tomamos
dt = 1/fs       # tiempo entre muestras consecutivas

t = np.arange(0, 1, dt)     # muestras

onda = (A[0] * np.sin(2*np.pi*f[0]*t)
    + A[1] * np.sin(2*np.pi*f[1]*t) 
    + A[2] * np.sin(2*np.pi*f[2]*t))

transformada = DFT(onda)
n = np.arange(fs)

plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.plot(t, onda)
plt.title(f'Onda 3 senos')
plt.grid(True)

plt.subplot(1, 3, 2)
plt.stem(t, onda)
plt.title(f'Onda 3 senos')
plt.grid(True)

freq = np.arange(len(transformada)) * fs / len(transformada)
magnitud = np.abs(transformada)

plt.subplot(1, 3, 3)
plt.stem(freq, np.abs(transformada))

plt.stem(freq, magnitud)

for i in f:
    plt.text(freq[i], magnitud[i]+2, f'{freq[i]:.0f} Hz, {magnitud[i]:.1f}')

plt.title('Transformada')
plt.xlabel('Frecuencia [Hz]')
plt.grid(True)

plt.tight_layout()
plt.show()
```
